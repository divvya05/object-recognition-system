# CIFAR-10 dataset exploration

Load CIFAR-10 using TensorFlow/Keras, inspect its dimensions, and visualize a batch of labeled examples.

```python
from tensorflow.keras.datasets import cifar10

(x_train, y_train), (x_test, y_test) = cifar10.load_data()
```

```python
class_names = [
    "airplane", "automobile", "bird", "cat", "deer",
    "dog", "frog", "horse", "ship", "truck",
]

print(f"Training images: {x_train.shape}")
print(f"Training labels: {y_train.shape}")
print(f"Test images: {x_test.shape}")
print(f"Test labels: {y_test.shape}")
print(f"Pixel range: {x_train.min()} to {x_train.max()}")
```

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(2, 5, figsize=(12, 5))
for index, ax in enumerate(axes.flat):
    ax.imshow(x_train[index])
    ax.set_title(class_names[int(y_train[index, 0])])
    ax.axis("off")

fig.tight_layout()
plt.show()
```
